```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Titanic dataset load panrom
df = sns.load_dataset('titanic')

# First 5 rows and summary paakalam
print(df.head())
print(df.info())
```

```python
# Missing values count paaka
print("Missing values count:")
print(df.isnull().sum())

# 1. 'age' column-la missing values-ku median replace panrom
df['age'] = df['age'].fillna(df['age'].median())

# 2. 'embarked' column-la missing values-ku mode replace panrom
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])

# 3. High missing value (~77%) irukra 'deck' column-ah drop panrom
df.drop(columns=['deck'], inplace=True)

print("\nClean pannadhukku apram missing values:")
print(df.isnull().sum())
```

```python
# Fare column-ku IQR calculate panrom
Q1 = df['fare'].quantile(0.25)
Q3 = df['fare'].quantile(0.75)
IQR = Q3 - Q1
upper_limit = Q3 + (1.5 * IQR)

print(f"Original Max Fare: {df['fare'].max():.2f}")
print(f"Calculated Upper Limit: {upper_limit:.2f}")

# Outliers-ah upper limit value-ku cap panrom
df['fare'] = np.where(df['fare'] > upper_limit, upper_limit, df['fare'])

print(f"Capping-ku apram Max Fare: {df['fare'].max():.2f}")# 'sex' column-ah 0 and 1-nu convert panrom
df['sex'] = df['sex'].map({'male': 0, 'female': 1})

# Cleaned data structure check panrom
print(df[['survived', 'pclass', 'sex', 'age', 'fare', 'embarked']].head())
print("\nFinal Null Check:")
print(df.isnull().sum())# 'sex' column-ah 0 and 1-nu convert panrom
df['sex'] = df['sex'].map({'male': 0, 'female': 1})

# Cleaned data structure check panrom
print(df[['survived', 'pclass', 'sex', 'age', 'fare', 'embarked']].head())
print("\nFinal Null Check:")
print(df.isnull().sum())
```
